```python
def integral_T(function, n):
    result = 0
    intervall_size = 1 / n #h
    for i in range(n):
        x_i = i * intervall_size
        x_iPlus1 = x_i + intervall_size
        current_area = intervall_size * .5 * (function(x_i) + function(x_iPlus1))
        result = result + current_area
    return result

def integral_R(function, n):
    intervall_size = 1 / n #h
    result = 0
    for i in range(n):
        x_i = i * intervall_size
        current_area = intervall_size * function(x_i)
        result = result + current_area
    return result
```

### b)

```python
import math
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
def f(x):
    return math.exp(x)

true_value = math.exp(1) - math.exp(0)

# calculate values for different Ns and compare to actual value
h = []
err_trapez = []
err_box = []

for i in range(0, 10):
    n = 2**(i+1)
    trapez_value = integral_T(f, n)
    box_value = integral_R(f, n)
    h.append(1/n)
    err_trapez.append(true_value - trapez_value)
    err_box.append(true_value - box_value)

# plot
fig, ax = plt.subplots()
ax.loglog(h, np.abs(err_trapez), label="trapez", marker='o')
ax.loglog(h, np.abs(err_box), label="box", marker='o')
ax.set_xlabel("h")
ax.set_ylabel("error")
ax.legend()
```

```python
# calculate c = err/h²

c_trapez = []
c_box = []

for i in range(len(h)):
    c_trapez.append(abs(err_trapez[i]) / (h[i]*h[i]))
    c_box.append(abs(err_box[i]) / (h[i]*h[i]))

# plot
fig, ax = plt.subplots(2, layout="constrained")

ax[0].plot(h, c_trapez, marker='o')
ax[0].set_title("trapez")
ax[0].set_xlabel("h")
ax[0].set_ylabel("err/h^2")

ax[1].plot(h, c_box, marker='o')
ax[1].set_title("box")
ax[1].set_xlabel("h")
ax[1].set_ylabel("err/h^2")
```

For the trapezoidal rule, the error approximately follows $C\cdot h^2$ with $C\approx -0.14$. For the box rule, the error does not follow a parabola.

```python
# calculate c = err/h

c_box = []

for i in range(len(h)):
    c_box.append(abs(err_box[i]) / h[i])

fig, ax = plt.subplots()
ax.plot(h, c_box, marker='o')
ax.set_title("box")
ax.set_xlabel("h")
ax.set_ylabel("err/h")
```

The error of the box rule approximately follows a linear function $C \cdot h$ with $C=e\approx 0.8$

```python
def integral_T_with_errorestimate(function, n):
    result = integral_T(function, n)
    comparison = integral_T(function, int(2*n))
    error_estimate = 4/3 * (result - comparison)
    return result, error_estimate

def integral_R_with_errorestimate(function, n):
    result = integral_R(function, n)
    comparison = integral_R(function, int(2*n))
    error_estimate = 4/3 * (result - comparison)
    return result, error_estimate
```

```python
r_trapez = []
r_box = []

for i in range(10):
    n = 2**(i+1)
    trapez_value, trapez_errorestimate = integral_T_with_errorestimate(f, n)
    box_value, box_errorestimate = integral_R_with_errorestimate(f, n)
    err_trapez = true_value - trapez_value
    err_box = true_value - box_value
    r_trapez.append(err_trapez/trapez_errorestimate)
    r_box.append(err_box/box_errorestimate)
    

# plot
fig, ax = plt.subplots(2, layout='constrained')
ax[0].semilogx(h, np.abs(r_trapez), marker='o')
ax[1].semilogx(h, np.abs(r_box), label="box", marker='o')
ax[0].set_xlabel("h")
ax[1].set_xlabel("h")
ax[0].set_ylabel("err/E(h)")
ax[1].set_ylabel("err/E(h)")
ax[0].set_title("trapez")
ax[1].set_title("box")
```

### a)

If a function $f:\mathbb{R}$ is a straight line on a loglog-plot, this means that $\ln(f(\exp(x)))$ is an affine linear function, so $\ln(f(\exp(x))) = C\cdot x + b$ for some $C,b\in \mathbb{R}$. 

If $f(h)=C\cdot h^\alpha$, then:
$$
\ln(C\cdot\exp(h)^\alpha) 
= \alpha\cdot\ln(C\cdot\exp(h)) 
= \alpha \cdot (\ln(C) + \ln(\exp(h))
= \alpha \cdot \ln(C) + \alpha\cdot h
$$

For $N \mapsto C\cdot \exp(-b\cdot N)$ i would use *semilogy*, because when the y-scale is logarighmic, the exponential function looks like a straight line.

For $h \mapsto C\cdot \exp(-\frac{b}{h})$, I would scale the y-axis logarithmically and use $\frac{1}{h}$ for the x-axis

For $h \mapsto C\cdot \exp(-\frac{b}{h^2})$, I would scale the y-axis logarithmically and use $\frac{1}{h^2}$ for the x-axis
